```python
# STEP 5: PYTHON IMPLEMENTATION

# Defines a function in taking the category dictionaries (budgets, spending) as input
def calculate_budget_status(budgets, spending):
    results = {}    # Creates an empty dictionary as nothing is calculated yet (start empty)

# loop over whichever collection represents the complete set of things you want in your output, and treat the other collection as optional/lookup data to fill in details.

# Loops every categories (including spending) within budgets on at a time
# ONLY budget, because it is the only COMPLETE and authoritative list, while user might not log in to a spending for a certain category
    for category in budgets:
        budget_amount = budgets[category] # Look up the category

# Cant be 0 or else percentage calculation will be off and crash Python
        if budget_amount <= 0:
            raise ValueError(f"Budget for {category} must be greater than zero")

# Under FOR loop of budget beacuse it may not be complete and not iterated over and result of absent spending category may not appear in the final results
# Lookup amount spent
        spent_amount = spending.get(category, 0)      # default 0 if nothing logged yet to prevent crash or error message
        if spent_amount < 0:
            raise ValueError(f"Spending for {category} cannot be negative")

        # Calculate how much budget is left and percentage of budget used
        remaining = budget_amount - spent_amount
        percent_used = (spent_amount / budget_amount) * 100

        if spent_amount > budget_amount:
            status = "overbudget"
        elif percent_used >= 80:
            status = "close to limit"
        else:
            status = "on track"

        # Create a nested dictionary to store everything inside one simple category
        results[category] = {
            "spent": spent_amount,
            "remaining": remaining,
            "percent_used": percent_used,
            "status": status
        }

    return results

# Defines function with 3 input:
def summarise_overall(budgets, spending, income):
    if income < 0:
        raise ValueError("Income cannot be negative")

    total_budgeted = sum(budgets.values())  # SUM the total budget value
    total_spent = sum(spending.values())  # SUM the total spending value
    unspent_from_income = income - total_spent
    unused_budget = total_budgeted - total_spent

    return {
        "total_budgeted": total_budgeted,
        "total_spent": total_spent,
        "unspent_from_income": unspent_from_income,
        "unused_budget": unused_budget
    }
```

```python
# TESTING CODE

budgets = {
    "Groceries": 400, "Rent": 1200, "Transport": 150,
    "Entertainment": 100, "Subscription": 35, "Gym": 48, "Mobile data": 39
}
spending = {
    "Groceries": 470, "Rent": 300, "Transport": 60,
    "Entertainment": 115, "Subscription": 70, "Gym": 24, "Mobile data": 35
}

result = calculate_budget_status(budgets, spending)

# Loops over dicitionary when you need both key and value at the same time
for category, info in result.items():
    print(f"{category}: spent = {info['spent']}, remaining = {info['remaining']}, "
          f"percent = {round(info['percent_used'], 1)}%, status = {info['status']}")
```

```python
# SUMMARY TEST

income = 2500
summary = summarise_overall(budgets, spending, income)
print(f"Total budgeted: {summary['total_budgeted']}")
print(f"Total spent: {summary['total_spent']}")
print(f"Unspent from income: {summary['unspent_from_income']}")
print(f"Unused budget: {summary['unused_budget']}")
```

```python
# STEP 6: TESTING WITH VARIETY OF DATA (MEET R6, R5, R3)

# Stores returned dictionary in result
result = calculate_budget_status(budgets, spending)

# Result reaches into outer dictionary and pulls in groceries' inner dictionary, same goes to status (grabbing the status string)
# Assert checks that it equals to "overbudget", if it matches nothing happens. If NOT, Python will stop and raise "AssertionError"
assert result["Groceries"]["status"] == "overbudget"
assert result["Groceries"]["remaining"] == -70

assert result["Rent"]["status"] == "on track"
assert result["Rent"]["remaining"] == 900

assert result["Entertainment"]["status"] == "overbudget"
assert round(result["Entertainment"]["percent_used"], 1) == 115.0

assert result["Mobile data"]["status"] == "close to limit"
assert result["Mobile data"]["remaining"] == 4

# Calls your other function with the same budgets/spending, plus income (2500), and stores the result in a flat dictionary of the four overall totals.
summary = summarise_overall(budgets, spending, income)
assert summary["total_budgeted"] == 1972
assert summary["total_spent"] == 1074
assert summary["unspent_from_income"] == 1426
assert summary["unused_budget"] == 898

print("All normal-case tests passed!")
```

```python
# EDGE AND ERROR CASE

# Edge case: category with no spending logged yet defaults to 0
# Looping over budgets and using .get() on spending so missing categories dont crash

no_spend_result = calculate_budget_status({"Gym": 48}, {})  # Lodges budget for Gym as $48, but keep dictionary empty to simulate user has not yet lodged any spending
# These check the .get(category,0) to ensure that fallback does not cause an error
assert no_spend_result["Gym"]["spent"] == 0  # Spent defaults to 0
assert no_spend_result["Gym"]["remaining"] == 48 # Remaining equals back to 48 (48-0) and with 0% used
assert no_spend_result["Gym"]["status"] == "on track"


# Use "try" patter specifically instead of calling the function directly is to prevent the error to be completely raised and crash the whole program and prevents other test from running
# "Try" expects the crash , catch it safely, and verify if its the right one

# Error case: zero budget should raise ValueError

try:  # Attempt the code and if it raises a specific kind of error, catch it here instead of crashing the whole program

    calculate_budget_status({"Gym": 0}, {"Gym": 10})  # Recalls your function with this guard clause (if budget amount <= 0,..)
    # Runs when the function above did  NOT raise an error, meaning that the zero-budget check is broken
    # This line will force the test to fail, with a clear message explaining the reasoning of the fail
    assert False, "Expected a ValueError for zero budget, but none was raised"
except ValueError as e:   # This catches error if one is is raised, and will be stored in variable "e" so we can inspect
    # Converts the caught error to a string, exactly the same as the line below ("Budget for Gym must be greater than zero")
    assert str(e) == "Budget for Gym must be greater than zero"



# Error case: negative spending should raise ValueError
try:
    calculate_budget_status({"Gym": 48}, {"Gym": -10})
    assert False, "Expected a ValueError for negative spending, but none was raised"
except ValueError as e:
    assert str(e) == "Spending for Gym cannot be negative"



# Error case: negative income should raise ValueError
# Testing summarise overall, with an income (-100)
try:
    summarise_overall({"Gym": 48}, {"Gym": 24}, -100)
    assert False, "Expected a ValueError for negative income, but none was raised"
except ValueError as e:
    assert str(e) == "Income cannot be negative"

print("All edge/error-case tests passed!")
```
