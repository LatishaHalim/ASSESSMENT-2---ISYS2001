```python
# Defines a function in taking the category dictionaries (budgets, spending) as input
def calculate_budget_status(budgets, spending):
    results = {}    # Creates an empty dictionary as nothing is calculated yet (start empty)

# loop over whichever collection represents the complete set of things you want in your output, and treat the other collection as optional/lookup data to fill in details.

# Loops every categories (including spending) within budgets on at a time
# ONLY budget, because it is the only COMPLETE and authoritative list, while user might not log in to a spending for a certain category
    for category in budgets:
        budget_amount = budgets[category] # Look up the category

        # A message if the value box is empty (UPDATED)
        if budget_amount is None:
            raise ValueError(f"Budget for {category} is empty - please enter a number")

        # Cant be 0 or else percentage calculation will be off and crash Python
        if budget_amount <= 0:
            raise ValueError(f"Budget for {category} must be greater than zero")

# Under FOR loop of budget beacuse it may not be complete and not iterated over and result of absent spending category may not appear in the final results
# Lookup amount spent
        spent_amount = spending.get(category, 0)      # default 0 if nothing logged yet to prevent crash or error message

        # A message if the value box is empty (UPDATED)
        if spent_amount is None:
            raise ValueError(f"Spending for {category} is empty - please enter a number")


        if spent_amount < 0:
            raise ValueError(f"Spending for {category} cannot be negative")

        # Calculate how much budget is left and percentage of budget used
        remaining = budget_amount - spent_amount
        percent_used = (spent_amount / budget_amount) * 100

        if spent_amount > budget_amount:
            status = "overbudget"
        elif percent_used >= 80:
            status = "close to limit"
        else:
            status = "on track"

        # Create a nested dictionary to store everything inside one simple category
        results[category] = {
            "spent": spent_amount,
            "remaining": remaining,
            "percent_used": percent_used,
            "status": status
        }

    return results

# Defines function with 3 input:
def summarise_overall(budgets, spending, income):
    # UPDATED (To close the gap)
    if income is None:
        raise ValueError("Income is empty - please enter a number")
    if income < 0:
        raise ValueError("Income cannot be negative")

    total_budgeted = sum(budgets.values())  # SUM the total budget value
    total_spent = sum(spending.values())  # SUM the total spending value
    unspent_from_income = income - total_spent
    unused_budget = total_budgeted - total_spent

    return {
        "total_budgeted": total_budgeted,
        "total_spent": total_spent,
        "unspent_from_income": unspent_from_income,
        "unused_budget": unused_budget
    }
```

# SET UP GEMINI API

```python
# Loads Python and carries out request to let code make HTTP calls (communicating with API)
# Crucial, or else, code below wont work
import requests
# Imports Colab's userdata tool to access Colab's secret (ONLY works in Colab)
from google.colab import userdata

# Pulls your key from Colab's Secrets, so it never appears as plain text in your notebook
api_key = userdata.get('GEMINI_API_KEY')
# Sending request to API
url = "https://generativelanguage.googleapis.com/v1beta/models/gemini-flash-latest:generateContent"

# header = this is literally how Google's API expects the key to be sent, per their docs.
headers = {
    "x-goog-api-key": api_key,  # Authentication, API want you to proof who you are (Confirms and track usage against free tier)
    # Tells the server that the data sent is formatted as JSON (So it knows to how to correctly read what is sent)
    "Content-Type": "application/json"

}
# The body of your request
# Gemini's expected request structure: your message goes inside a nested parts list (Contents -> Parts -> Text)
payload = {  # A dictionary
    # Contents is a list (a conversation could have multiple returns/response)
    "contents": [
        {
            # Parts is a list (A meessage could have different forms e.g. text, image,..)
            "parts": [
                # Text is a string holding the actual message
                {"text": "Say hello and confirm you're working, in one sentence."}
            ]
        }
    ]
}
# Sends HTTP POST request to URL to retrieve data
# Attaching header and convert payload dictionary to JSON text AUTOMATICALLY
response = requests.post(url, headers=headers, json=payload)
# Whatever Google server sends back is stored in response
# it contains status info, headers, and the actual reply, but not in an easily readable format yet.
data = response.json() # (.json) converts the raw HTTP response into a Python dictionary you can work with.
print(data)
```

# MEETING R1

```python
# To get just the sentence out instead of nested dictionaries and list
#  --> Need to navigate by : data["candidates"][0]["content"]["parts"][0]["text"]

import time

# Helper function:
# defines a function with 3 parameters user_message (what you want to ask), api_key (your credentials), system_instruction=None (The =None makes it optional, if you call ask_gemini("hello", api_key) without a third argument, system_instruction just defaults to None rather than causing an error demanding a missing argument.)
# Sends a message to Gemini and returns just the reply text.
# Automatically retries a couple of times if the model is temporarily overloaded (503 error), since this is common on the free tier.
def ask_gemini(user_message, api_key, system_instruction=None, retries=5):

    # Now living inside the function, so it's automatically used every time you call ask_gemini, rather than you retyping it each time.
    url = "https://generativelanguage.googleapis.com/v1beta/models/gemini-flash-latest:generateContent"

    headers = {
        "x-goog-api-key": api_key,  # now api_key refers to whatever was passed in as an argument when the function was called NOT a fixed value, so this function works with any key you give it.
        "Content-Type": "application/json"
    }
    # user_message (the function's argument) makes the function reusable for any message, not just fixed one
    payload = {
        "contents": [{"parts": [{"text": user_message}]}]
    }

    # Check whether a syestem_instruction is provided
    # If system_instruction treats none (the default) as falsy, this means that the block is skipped if nothing is passed in
    # If something is passed like your persona text this line adds an extra key "systemInstruction", to the payload dictionary, structured the way Gemini expects a system instruction (same parts/text nesting as a regular message, just under a different top-level key).
    # Meeting R1
    if system_instruction:
        payload["systemInstruction"] = {"parts": [{"text": system_instruction}]}

    for attempt in range(retries):
        response = requests.post(url, headers=headers, json=payload)

        if response.status_code == 200:
            data = response.json()
            return data["candidates"][0]["content"]["parts"][0]["text"]

        elif response.status_code == 503:
            time.sleep(5)  # wait 2 seconds, then let the loop try again

        else:
            raise ValueError(f"Gemini API error {response.status_code}: {response.text}")

    raise ValueError("Gemini API unavailable after multiple retries — try again shortly")
```

# GROUND THE ASSISTANT REAL DATA (R6)

PLAN: before sending the user's question to Gemini, we build a short summary of their real budget numbers and include it in the prompt, so the reply references actual figures instead of generic advice.

```python
# Where R1 (Persona) and R2 (Grounded in real data) is connected

# It calls your two existing functions internally, this function doesn't recalculate anything itself, it just formats what they already produced into readable sentences.
def build_budget_summary(budgets, spending, income):
   #   Turns the budget status and summary data into a short text block that can be included in a prompt to Gemini.
    status = calculate_budget_status(budgets, spending)
    summary = summarise_overall(budgets, spending, income)


    lines = ["Here is the user's current budget situation:"]
    for category, info in status.items():
        lines.append(   # Builds up a list of text lines one at a time, one per category, rather than trying to construct one giant string directly.
            f"- {category}: budgeted ${budgets[category]}, spent ${info['spent']}, "
            f"remaining ${info['remaining']} ({round(info['percent_used'], 1)}% used, {info['status']})"
        )

    lines.append(
        f"Overall: income ${income}, total budgeted ${summary['total_budgeted']}, "
        f"total spent ${summary['total_spent']}, unspent from income ${summary['unspent_from_income']}."
    )

    # Takes the list of separate line-strings and glues them into one string, with a newline character (\n) between each. This is how you turn a list of lines into a proper multi-line block of text.
    return "\n".join(lines)
```

```python
# WIRING INTO GEMINI CALLS
# This wraps exsiting ask_gemini without changing it, instead making the message better before calling it

#  Asks Gemini a question, grounding the reply in the user's real budget data.
def ask_gemini_grounded(user_message, api_key, persona, budgets, spending, income):

    budget_context = build_budget_summary(budgets, spending, income)

    full_message = (
        f"{budget_context}\n\n"
        f"The user asks: {user_message}\n\n"
        # Without it, the model might still give generic advice even with the data sitting right there in the prompt.
        f"Answer using their actual numbers above where relevant."
    )

    return ask_gemini(full_message, api_key, persona)
```

```python
# Loads the (gr)adio library
import gradio as gr

# Function 1: check budget status from Gradio inputs

# Defines a function with 15 parameters: income, 7 category budget (_B), 7 catergory spending (_S)
# Gradio will automaticall call it, filling each parameter from the matching input box, in the order they're listed
def check_budget(income, groceries_b, rent_b, transport_b, entertainment_b, subscription_b, gym_b, mobile_b,
                  groceries_s, rent_s, transport_s, entertainment_s, subscription_s, gym_s, mobile_s):

    # Takes the 7 separate _b parameters and reassembles them into the same dictionary shape your calculate_budget_status function expects
    # Gradio hands you individual values per box, not a ready-made dictionary. This line is the bridge between "Gradio's world" (separate numbers) and "your function's world" (a dictionary)
    budgets = {
        "Groceries": groceries_b, "Rent": rent_b, "Transport": transport_b,
        "Entertainment": entertainment_b, "Subscription": subscription_b,
        "Gym": gym_b, "Mobile data": mobile_b
    }
    spending = {
        "Groceries": groceries_s, "Rent": rent_s, "Transport": transport_s,
        "Entertainment": entertainment_s, "Subscription": subscription_s,
        "Gym": gym_s, "Mobile data": mobile_s
    }


    # Related to R3 and R4
    # If either raises a ValueError (a zero budget, negative spending, or (after your fix) an empty box) this catches it here, at the Gradio layer, and immediately returns a friendly warning string instead of letting the error crash the whole interface
    try:
        status = calculate_budget_status(budgets, spending)
        summary = summarise_overall(budgets, spending, income)
    except ValueError as e:
        return f"⚠️ Input error: {e}"

    lines = []
    for category, info in status.items():
        lines.append(
            f"**{category}**: spent ${info['spent']} of ${budgets[category]} "
            f"— {round(info['percent_used'],1)}% used — remaining ${info['remaining']} — **{info['status']}**"
        )

    lines.append("\n---")  # Adds a horizontal divider line '---' (a visual break between per category break down and the overall totals)
    # Four more lines, pulling from summary dictionary returned by summarise overall
    lines.append(f"**Total budgeted:** ${summary['total_budgeted']}")
    lines.append(f"**Total spent:** ${summary['total_spent']}")
    lines.append(f"**Unspent from income:** ${summary['unspent_from_income']}")
    lines.append(f"**Unused budget:** ${summary['unused_budget']}")
    # Joins all accumulated line into one string with a blank line (\n\n two new line) between each
    return "\n\n".join(lines)


# Function 2: chat with the grounded assistant
# Same as functions in the cell above with additional user_question
def chat_with_assistant(user_question, income, groceries_b, rent_b, transport_b, entertainment_b,
                         subscription_b, gym_b, mobile_b, groceries_s, rent_s, transport_s,
                         entertainment_s, subscription_s, gym_s, mobile_s):
# Same as check_budget with two dictionaries --> Gradio calls out each function individually with its own set of current box values
    budgets = {
        "Groceries": groceries_b, "Rent": rent_b, "Transport": transport_b,
        "Entertainment": entertainment_b, "Subscription": subscription_b,
        "Gym": gym_b, "Mobile data": mobile_b
    }
    spending = {
        "Groceries": groceries_s, "Rent": rent_s, "Transport": transport_s,
        "Entertainment": entertainment_s, "Subscription": subscription_s,
        "Gym": gym_s, "Mobile data": mobile_s
    }

    # .strip removes leading white space from the question
    if not user_question.strip():
        return "Please type a question first."
    # Calls existing grounding function
    # If it succeeds if will return in assistant's response
    # If fails shows error message instead of crashin
    try:
        return ask_gemini_grounded(user_question, api_key, persona_instruction, budgets, spending, income)
    except ValueError as e:
        return f"⚠️ The assistant couldn't respond right now: {e}"


# Build the interface

# gr.blocks = Gradio's flexible layout builder
# Blocks allow you to arrange multiple inputs, outputs, and buttons freely, which is why it's suited to an app with two separate features (budget checker and chat)
# The with ... as demo: pattern means "everything indented below belongs to this interface," and the whole thing is stored in a variable called demo
# title= sets the browser tab title.
with gr.Blocks(title="Student Budget Assistant") as demo:
    gr.Markdown("# 💰 Student Budget Assistant"). # The # in Markdown makes a large heading,
    gr.Markdown("Enter your income and budgets, check your status, then ask the assistant a question.")

    # Creates a number input box
    # This entire component is stored in the variable income_input, which is how it gets referenced later when wiring up the button
    # label= is the text shown above the box
    # value=2500 is the default number shown when the interface first loads (this is where your worked-example default of $2500 comes from)
    income_input = gr.Number(label="Monthly Income ($)", value=2500)

    # gr.Row() is a layout container, everything inside it is arranged horizontally, side by side, instead of stacked vertically.
    # This groups four budget input boxes into one visual row
    # Each box is created the same way as income_input, each with its own default matching your worked example
    gr.Markdown("### Budgets per category ($)")
    with gr.Row():
        groceries_b = gr.Number(label="Groceries budget", value=400)
        rent_b = gr.Number(label="Rent budget", value=1200)
        transport_b = gr.Number(label="Transport budget", value=150)
        entertainment_b = gr.Number(label="Entertainment budget", value=100)
    # A second row for the remaining three budget categories
    # Split into two rows purely so the layout doesn't stretch too wide on screen.
    with gr.Row():
        subscription_b = gr.Number(label="Subscription budget", value=35)
        gym_b = gr.Number(label="Gym budget", value=48)
        mobile_b = gr.Number(label="Mobile data budget", value=39)

    # Same pattern repeated for the 7 spending boxes, defaulting to 0 (no spending logged yet) rather than your worked-example figures
    gr.Markdown("### Actual spending so far ($)")
    with gr.Row():
        groceries_s = gr.Number(label="Groceries spent", value=0)
        rent_s = gr.Number(label="Rent spent", value=0)
        transport_s = gr.Number(label="Transport spent", value=0)
        entertainment_s = gr.Number(label="Entertainment spent", value=0)
    with gr.Row():
        subscription_s = gr.Number(label="Subscription spent", value=0)
        gym_s = gr.Number(label="Gym spent", value=0)
        mobile_s = gr.Number(label="Mobile data spent", value=0)

    # Creates a clickable button "Check My Budget", and an empty Markdown display area to hold the result once it's calculated
    # gr.Markdown() with nothing inside starts blank, it only shows content once something is returned into it.
    check_button = gr.Button("Check My Budget")
    budget_output = gr.Markdown()

    # Bundles all 15 input-box variables into one list, in the exact same order as check_budget's parameters.
    # This ordering is critical — Gradio matches list position to function parameter position, not by name, so if this order didn't match check_budget's parameter order, values would land in the wrong variables inside the function (e.g. rent's value ending up treated as transport's)
    budget_inputs = [income_input, groceries_b, rent_b, transport_b, entertainment_b, subscription_b, gym_b, mobile_b,
                      groceries_s, rent_s, transport_s, entertainment_s, subscription_s, gym_s, mobile_s]
    # This is the actual wiring
    # When check_button is clicked, call check_budget, passing in the current values of everything in budget_inputs, and put whatever it returns into budget_output
    # Nothing happens until the button is clicked, that's what makes this event-driven rather than running automatically on page load
    check_button.click(fn=check_budget, inputs=budget_inputs, outputs=budget_output)

    # The second feature's component
    # A heading, a text input box (gr.Textbox, different from gr.Number since a question is free text, not a number)
    # Placeholder= shows greyed-out example text that disappears once the user starts typing
    # "Ask" button, and another empty output area
    gr.Markdown("### Ask the Assistant")
    question_input = gr.Textbox(label="Your question", placeholder="e.g. Am I doing okay this month?")
    ask_button = gr.Button("Ask")
    chat_output = gr.Markdown()


    chat_inputs = [question_input] + budget_inputs
    # Same wiring pattern as the check button, connecting the "Ask" button to chat_with_assistant.
    ask_button.click(fn=chat_with_assistant, inputs=chat_inputs, outputs=chat_output)
# This is outside the with gr.Blocks(...) as demo: block (notice it's not indented)
# It takes the fully-built interface stored in demo and actually starts it running, which is what produced the gradio
# Live link and the embedded panel you tested against
# Without this line, the interface would be defined but never actually opened
demo.launch()
```
