THE SIX-STEP METHOD IN PLANNING

1. Understanding the problem:

It is easy for international students to lose track of mindful spending when source of income is given and freedom of spending is present. This financial assistant helps regulate financial spendings and incomes in order to keep in track of users financial goal and stability.




2. Identify INPUTS and OUTPUTS:

* Data field will include: Date, Amount, Category, Description (on track, close to limit, overbudget), and Free-text section (for financial advices from assistant)

> *Expected Input:*

--> Date

--> Monthly Income (float)

--> Budget limits per cateogry (Value)

--> Spending amount (float)

--> Questions to ask budgeting assistant

>  *Expected Output:*

--> Financial status per category:
* amount spent
* Remaining budget
* % of budget used
* Flag on whether spending (on track, close to limit, overbudget)

--> Financial summary of overall total spending and total budget and income received

--> Financial advice/reply from budgeting assistant
